# How to Rename Columns in Polars

## Establishing a Baseline

```python
import polars as pl

data = pl.read_csv('../datasets/employees.csv')
data.head(1)
```

## Renaming After Import


### Using a List

```python
data.columns = ['A', 'B', 'C','D', 'E', 'F', 'G', 'H']
data.head(1)
```

### Using the Rename Function

```python
data.rename(mapping={'A': 'Emp_ID'
                    ,'B': 'First_Name'
                    , 'C': 'Last_Name'
                    , 'D': 'Email'
                    , 'E': 'Job_Role'
                    , 'F': 'Date_Of_Birth'
                    , 'G': 'Hire_Date'
                    , 'H': 'Salary'}).head(1)
```

## Renaming on Import

### Using the New Columns Parameter

```python
data = pl.read_csv('../datasets/employees.csv',new_columns=['ID'
                                                            ,'FirstName'
                                                            ,'LastName'
                                                            ,'Email'
                                                            ,'Job'
                                                            ,'DOB'
                                                            ,'HireDate'
                                                            ,'Salary'])
data.head(1)
```

### Using a Schema

```python
import polars_extensions as plx
schema = plx.read_schema('schema.json')
schema
```

```python
data = pl.read_csv('../datasets/employees.csv',schema=schema)
data.head(1)
```

### Using DataFramely Model

```python
import dataframely as dy

class Employees(dy.Schema):
    EmployeeId = dy.Integer(primary_key=True)
    FirstName = dy.String(nullable=False)
    LastName = dy.String(nullable=False)
    Email = dy.String(nullable=False)
    JobTitle = dy.String(nullable=False)
    DateOfBirth = dy.Date(nullable=False)
    DateOfHire = dy.Date(nullable=False)
    Salary = dy.Float(nullable=False)

dataframely_schema = Employees.polars_schema()
dataframely_schema 
```

```python
data = pl.read_csv('../datasets/employees.csv',schema=dataframely_schema )
data.head(1)
```
