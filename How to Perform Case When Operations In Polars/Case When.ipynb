## Case When

```python
import polars as pl

data = pl.read_csv('../datasets/credit_card_transactions.csv')
data
```

```python
data.with_columns(
                pl.when(pl.col('amount') > 500).then(pl.lit('High'))
                .when(pl.col('amount') > 100).then(pl.lit('Medium'))
                .otherwise(pl.lit('Low')).alias('spend_category'))
```

```python
high = pl.col('amount') > 400
medium = pl.col('amount') > 100

data = data.with_columns(
                pl.when(high).then(pl.lit('High'))
                .when(medium).then(pl.lit('Medium'))
                .otherwise(pl.lit('Low')).alias('spend_category'))
data
```

```python
data.group_by('spend_category').agg(pl.col('amount').count().alias('transaction_count')
                                    ,pl.col('amount').sum().alias('total_amount'))
```
