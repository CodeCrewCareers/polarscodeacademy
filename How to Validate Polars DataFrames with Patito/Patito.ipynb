## The Model

```python
import polars as pl
import patito as pt
from typing import Literal, Optional

class Request(pt.Model):
    requestId: int = pt.Field(unique=True)
    requestor: str
    requestType: Literal['Add', 'Edit', 'Update', 'Delete']
    description: Optional[str]
    requestDate: str = pt.Field(dtype=pl.Date)
```

```python
Request.valid_dtypes
```

## The Data

```python
df = pl.read_csv('../datasets/requests.csv')
df
```

## Validate the Data

```python
try:
    Request.validate(df)
    print('Validation successful')
except pt.DataFrameValidationError as exc:
    print(exc)
```
