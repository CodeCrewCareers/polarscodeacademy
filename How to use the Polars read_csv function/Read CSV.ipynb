# Read CSV

## Defaults

```python
import polars as pl
data = pl.read_csv('../datasets/employees.csv')
data.head()
```

## Schema Overrides

```python
data = pl.read_csv('../datasets/employees.csv', schema_overrides={'salary':pl.Float64,'date_of_birth':pl.Date,'date_of_hire':pl.Date})
data.head()
```

## Try Parse Dates

```python
data = pl.read_csv('../datasets/employees.csv',try_parse_dates=True)
data.head()
```

## Has Header

```python
data = pl.read_csv('../datasets/employees.csv', has_header=False)
data.head()
```

## New Columns

```python
col_names = ['EID','FName','LName','Email','Position','DOB','HireDate','Compensation']
data = pl.read_csv('../datasets/employees.csv', has_header=False, new_columns=col_names)
data.head()
```

```python
data = pl.read_csv('../datasets/employees.csv', new_columns=col_names)
data.head()
```

## Separator

```python
data = pl.read_csv('../datasets/employees.tsv')
data.head()
```

```python
data = pl.read_csv('../datasets/employees.tsv',separator='\t',)
data.head()
```
