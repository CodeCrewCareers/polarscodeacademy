# Hashing Senstive Data

```python
import polars as pl
```

```python
data = pl.read_csv('../Datasets/fake_PII.csv')
data.head()
```

## hash

```python
data.with_columns(pl.col('Name').hash().alias('Name_Hash'),
                  pl.col('Address').hash().alias('Address_Hash')
                  )
```

## Seed(s)

```python
data.with_columns(pl.col('Name').hash(seed=42).alias('Name_Hash'),
                  pl.col('Address').hash(seed=42).alias('Address_Hash'))
```

## Seed Stacking

```python
data.with_columns(pl.col('Name').hash(seed=42, seed_1=3).alias('Name_Hash'),
                  pl.col('Address').hash(seed=42, seed_1=3).alias('Address_Hash'))
```

## Important Note!

This implementation of hash does not guarantee stable results across different Polars versions. Its stability is only guaranteed within a single version.
