# API to Polars DataFrame

## Import Secrets

```python
import json

secrets = json.load(open('secrets.config'))
apikey = secrets['API_KEY']
apitoken = secrets['API_TOKEN']
id = secrets['id']
```

## API Call (Getting All Cards on a Board)

```python
import requests
import json


url = f"https://api.trello.com/1/boards/{id}/cards"

headers = {
  "Accept": "application/json"
}

query = {
  'key': apikey,
  'token': apitoken
}

response = requests.request(
   "GET",
   url,
   headers=headers,
   params=query
)
```

## Load JSON into DataFrame

```python
import polars as pl 

data = pl.json_normalize(json.loads(response.text))
data
```

## Drilling into Card Checklists

```python
import requests

url = "https://api.trello.com/1/checklists/691945971e7f86c316c5e5a5"

response = requests.request(
   "GET",
   url,
   params=query
)
```

```python
import polars as pl 

checklist = pl.json_normalize(json.loads(response.text))
checklist.select(pl.col('checkItems')).explode('checkItems').unnest('checkItems')
```
