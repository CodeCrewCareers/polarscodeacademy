# MySQL to Polars DataFrame

## Using ConnectorX

```python
import polars as pl

uri = "mysql://devuser:devpass@127.0.0.1:3306/devdb"
query = "SELECT * FROM users"
pl.read_database_uri(query=query, uri=uri)
```

## SQLAlchemy

```python
import polars as pl
from sqlalchemy import create_engine

conn = create_engine("mysql+pymysql://devuser:devpass@127.0.0.1:3306/devdb")
query = "SELECT * FROM opportunities"

df = pl.read_database(query=query, connection=conn.connect())
df
```
