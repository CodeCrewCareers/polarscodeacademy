# Numbers Stored as Text

## Integers

```python
import polars as pl
df = pl.DataFrame({"number": ["544"]})
df.with_columns(parsed=pl.col("number").str.to_integer(base=10, strict=False))
```

## Decimals

![alt text](../../Resources/Precision.svg)

### Uniform Precision & Scale

```python
import polars as pl

df = pl.DataFrame({"decimals": ["6.4", "5.4","4.5"]})
df.with_columns(parsed=pl.col("decimals").str.to_decimal())
```

### Mixed Precision & Scale

```python
df = pl.DataFrame({"decimals": ["-6.433223354332", "5.4877","45.54443434"]})
df.with_columns(parsed=pl.col("decimals").str.to_decimal())
```

### Inferring Scale


```python
df = pl.DataFrame({"decimals": ["-6.43", "5.4877","45.54443434"]})
df.with_columns(parsed=pl.col("decimals").str.to_decimal(inference_length=1))
```

### Max Precision (38)


```python
df = pl.DataFrame({"decimals": [".88999876567899879878987789998789877665","74.9486478958473625"]})
df.with_columns(parsed=pl.col("decimals").str.to_decimal())
```
