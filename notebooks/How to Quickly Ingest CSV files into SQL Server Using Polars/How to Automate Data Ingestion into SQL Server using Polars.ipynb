# How to Quickly Ingest CSV files into SQL Server Using Polars

```python
import requests

url = "https://raw.githubusercontent.com/CodeCrewCareers/polarscodeacademy/refs/heads/main/datasets/employees.csv"
response = requests.get(url)
response.raise_for_status()

with open("employees.csv", "wb") as f:
    f.write(response.content)
```

```python
import polars as pl

df = pl.read_csv("employees.csv")
df.head()
```

## Connection String Tips
* Make sure you have a Driver Installed
    * https://learn.microsoft.com/en-us/sql/connect/odbc/download-odbc-driver-for-sql-server?view=sql-server-ver17
* Carefully check your port and server name



```python
import sqlalchemy

uri = f"mssql+pyodbc://localhost/Live?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"

engine = sqlalchemy.create_engine(uri, fast_executemany=True)
engine
```

```python
df.write_database(table_name='dbo.Employees',connection=engine,if_table_exists='replace')
```
