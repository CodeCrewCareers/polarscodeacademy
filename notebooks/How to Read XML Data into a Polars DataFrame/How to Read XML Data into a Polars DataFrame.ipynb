# Read XML

## Defaults

```python
import polars_extensions as plx
data = plx.read_xml(xml_input='products.xml')
data
```

## Flatten

```python
import polars_extensions as plx
data = plx.read_xml(xml_input='products.xml',flatten=False)
data
```

## Include Attributes

```python
import polars_extensions as plx
data = plx.read_xml(xml_input='products.xml',include_attributes=False)
data
```

## Record Path

```python
import polars_extensions as plx
data = plx.read_xml(xml_input='products.xml',record_path='catalog.product.features',include_attributes=False)
data
```

```python
import polars_extensions as plx
data = plx.read_xml(xml_input='products.xml',record_path='catalog.product',include_attributes=False,flatten=False)
data
```

## Text Input

```python
import polars_extensions as plx
xml_data = '''
<catalog>
    <product sku="A123" available="true">
        <name>Mechanical Keyboard</name>
        <price currency="USD">129.99</price>
        <features>
            <feature>RGB Lighting</feature>
            <feature>Hot-swappable switches</feature>
            <feature>Aluminum frame</feature>
        </features>
    </product>

    <product sku="B456" available="false">
        <name>Noise Cancelling Headphones</name>
        <price currency="USD">299.00</price>
        <features>
            <feature>ANC</feature>
            <feature>Bluetooth 5.0</feature>
        </features>
    </product>
</catalog>
'''

df = plx.read_xml(xml_input=xml_data)
df
```
