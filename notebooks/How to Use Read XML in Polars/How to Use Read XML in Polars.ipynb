# How to Use Read XML in Polars

## Defaults

```python
import polars_extensions as plx
import polars as pl

data = plx.read_xml('../../datasets/rss.xml')
data
```

## Flatten

```python
data = plx.read_xml('../../datasets/rss.xml',flatten=False)
data
```

## Record Path

```python
data = plx.read_xml('../../datasets/rss.xml' ,record_path='rss.channel.item')
data
```

```python
data = plx.read_xml('../../datasets/rss.xml' ,record_path='rss.channel.item',flatten=False)
data
```

```python
data.drop('item.category').explode('item.title.text','item.link.text','item.description.text','item.pubDate.text')
```

## Include Attributes

```python
data = plx.read_xml('../../datasets/rss.xml',include_attributes=False)
data
```
